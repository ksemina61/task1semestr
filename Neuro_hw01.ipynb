```python
# перекопировала данные из тетрадки autograd
import numpy as np
import torch
```

```python
batch_size = 64
input_size = 3 # кол-во признаков (x = 3)
hidden_size = 2 # скрытый слой (2 нейрона)
output_size = 1 # 1 нейрон
```

```python
dtype = torch.float
device = torch.device("cpu")
```

```python
# Тут создается рандмные input and output data
x = torch.randn(batch_size, input_size, device=device, dtype=dtype)
y = torch.randn(batch_size, output_size, device=device, dtype=dtype)
# Также рандомно инизиализируются весы
w1 = torch.randn(input_size, hidden_size, device=device, dtype=dtype)
w2 = torch.randn(hidden_size, output_size, device=device, dtype=dtype)

# Добавлю смещение b1, b2
b1 = torch.randn(hidden_size, device = device,dtype=dtype)
b2 = torch.randn(output_size, device = device, dtype=dtype)
learning_rate = 1e-6
```

```python
for t in range(500):
    # Первый слой, прямой проход
    h_1 = x.mm(w1) + b1 # умножение входных данных x на веса w1 и добавления смещения b1
    h_relu = h_1.clamp(min=0)  # функция ReLU к выходу 1 слоя, отрицательные значения теперь зануляются
    out = h_relu.mm(w2) + b2 # выход нейронной сети, то есть умножается выход ReLU-слоя h_relu на веса w2 и добавляется смещения b2

    # Функция потерь
    loss = ((out - y) ** 2).sum().item()

    # Если вычислять здесь градиент вручную, что должно получится (вычисление градиента по выходу):
    # dL/dout = 2 * (out - y), где  L = sum((out - y)^2)
    dloss = 2 * (out - y)

    # Обратный проход
    # Второй слой
    # нужно использовать цепное правило; (dL/dout) = dloss
    grad_w2 = h_relu.t().mm(dloss) # вручную: dL/dw2 = (dL/dout) * (dout/dw2) = dloss * h_relu(dL/dout * h_relu)
    grad_b2 = dloss.sum(dim=0) # вручную: dL/db2 = (dL/dout) * (dout/db2) = dloss * 1

    # Backpropagation (обратное распространение ошибки)
    # dout/dh_relu =  w2
    grad_hrelu = dloss.mm(w2.t()) # если вычислять вручную: dL/dh_relu = (dL/dout) * (dout/dh_relu) = dloss * (dout/dh_relu)
    grad_hrelu[h_1 < 0] = 0

    # Градиенты для первого слоя w1, b1
    # dL/dh_relu = grad_hrelu
    grad_w1 = x.t().mm(grad_hrelu) # вручную: dL/dw1 = (dL/dh_relu) * (dh_relu/dh_1) * (dh_1/dw1) или просто dL/dw1 = dL/dh_relu * x
    grad_b1 = grad_hrelu.sum(dim=0) # вручную: dL/db1 = (dL/dh_relu) * (dh_relu/dh_1) * (dh_1/db1) или просто dL/db1 = dL/dh_relu

    # Обновление весов
    w1 -= learning_rate * grad_w1
    w2 -= learning_rate * grad_w2
    b1 -= learning_rate * grad_b1
    b2 -= learning_rate * grad_b2

    if t % 100 == 99:
        print(f'Loss on iteration {t} = {loss}')
        print(f'grad_w1:{grad_w1}')
        print(f'grad_w2:{grad_w2}')
        print(f'grad_b1:{grad_b1}')
        print(f'grad_b2:{grad_b2}\n\n')
```

```python
# поняла, что можно не расписывать постоянно w1 = torch.randn(input_size, hidden_size, device=device, dtype=dtype, requires_grad=True)
# поэтому использую метод requires_grad_() - добавляет уже существующие данные
x_auto = x
y_auto = y
w1_auto = w1.requires_grad_()
w2_auto = w2.requires_grad_()
b1_auto = b1.requires_grad_()
b2_auto = b2.requires_grad_()
```

```python
for t in range(500):
    # Прямой проход
  h_1 = x_auto.mm(w1_auto) + b1_auto
  h_relu = h_1.clamp(min=0)
  out = h_relu.mm(w2_auto) + b2_auto

    # Функция потерь
  loss_auto = ((out - y_auto) ** 2).sum()
  if t % 100 == 99:
    print(f'Loss on iteration {t} = {loss_auto.item()}')
    print(f'grad_w1:{w1_auto.grad}')
    print(f'grad_w2:{w2_auto.grad}')
    print(f'grad_b1:{b1_auto.grad}')
    print(f'grad_b2:{b2_auto.grad}\n\n')

    # считает градиенты для весов
  loss_auto.backward()

    # Обновление весов
  with torch.no_grad():
    w1_auto -= learning_rate * w1_auto.grad
    w2_auto -= learning_rate * w2_auto.grad
    b1_auto -= learning_rate * b1_auto.grad
    b2_auto -= learning_rate * b2_auto.grad

# сбрасываем накопленные градиенты
    w1_auto.grad.zero_()
    w2_auto.grad.zero_()
    b1_auto.grad.zero_()
    b2_auto.grad.zero_()
```

Формула:

В двух измерениях формула выглядит так: d(p, q) = √(p1 − q1)² + (p2 − q2)².

В трёх измерениях для точек, заданных их декартовыми координатами, расстояние равно d(p, q) = √(p1 − q1)² + (p2 − q2)² + (p3 − q3)²
